```python
# Tenta ler o arquivo recém-criado
data_file = P("base_dados_ecommerce.csv")

if not data_file.exists():
    raise FileNotFoundError("Arquivo não encontrado. Rode a célula de criação de dados primeiro.")

# Carrega o CSV para o DataFrame (nome alterado para ecom_data)
ecom_data = pd_data.read_csv(data_file)

# Informações de volume
print(f"O DataFrame possui {ecom_data.shape[0]} observações e {ecom_data.shape[1]} atributos.")

# Exibe as primeiras entradas
show(ecom_data.head(3)) # Mostra as 3 primeiras linhas
```

```python
# Visualização dos Tipos de Dados e Preenchimento
print("### Estrutura do DataFrame ###")
ecom_data.info(verbose=False) # Mostra tipos de dados e contagem de entradas

# Estatísticas Descritivas
print("\n### Estatísticas Resumidas (Tudo Incluído) ###")
show(ecom_data.describe(include='all')) # Inclui categóricas
```

```python
# Identifica a coluna categórica para One-Hot Encoding (OHE)
CATEGORICAL_COL = 'product_name'

# Realiza a codificação OHE (get_dummies) e elimina a primeira categoria para evitar multicolinearidade
processed_data = pd_data.get_dummies(ecom_data, columns=[CATEGORICAL_COL], drop_first=True)

# Exibe a nova estrutura após a codificação
print("Dimensões do DataFrame após OHE:", processed_data.shape)
print("\nPrimeiras linhas do DataFrame processado:")
show(processed_data.head())
```

```python
# Define a variável alvo (y) e as features (X)
TARGET_VARIABLE = 'purchased'
feature_set = processed_data.drop(TARGET_VARIABLE, axis=1) # Todas as colunas exceto a alvo
target_output = processed_data[TARGET_VARIABLE] # A coluna alvo

# Divide em conjuntos de treino e teste (70% Treino, 30% Teste)
# Usamos a semente definida anteriormente (42) para reprodutibilidade e 'stratify' para manter a proporção da variável alvo
X_train, X_test, y_train, y_test = tts(
    feature_set, target_output, test_size=0.3, random_state=RANDOM_SEED, stratify=target_output
)

# Confirmação dos tamanhos dos conjuntos
print(f"Conjunto de Treino (X, y): {X_train.shape} e {y_train.shape}")
print(f"Conjunto de Teste (X, y): {X_test.shape} e {y_test.shape}")
```

```python
# Inicializa o classificador de Árvore de Decisão
model_tree = DTC(max_depth=4, random_state=RANDOM_SEED) # Limita a profundidade para evitar overfitting

# Treina o modelo
model_tree.fit(X_train, y_train)
print("Modelo de Árvore de Decisão treinado com sucesso.")

# Faz previsões no conjunto de teste
test_predictions = model_tree.predict(X_test)

# === Avaliação do Desempenho ===
print("\n### Métricas de Avaliação no Conjunto de Teste ###")
print(f"Acurácia: {accuracy_score(y_test, test_predictions):.4f}")
print(f"Precisão: {precision_score(y_test, test_predictions):.4f}")
print(f"Recall (Revocação): {recall_score(y_test, test_predictions):.4f}")
print(f"F1-Score: {f1_score(y_test, test_predictions):.4f}")

# Exibe a Matriz de Confusão
print("\nMatriz de Confusão (Linhas: Real, Colunas: Previsto):")
show(pd_data.DataFrame(confusion_matrix(y_test, test_predictions), index=['Real: 0', 'Real: 1'], columns=['Previsto: 0', 'Previsto: 1']))
```

```python
# Cria a visualização gráfica da Árvore de Decisão
pplt.figure(figsize=(18, 12)) # Ajuste o tamanho da figura para melhor visualização
tree_viz(
    model_tree,
    feature_names=X_train.columns.tolist(),
    class_names=['Não Comprado (0)', 'Comprado (1)'],
    filled=True,
    rounded=True,
    impurity=False, # Não exibe a impureza, melhora a clareza
    proportion=True, # Exibe as proporções de amostras
    fontsize=8
)
pplt.title("Visualização da Árvore de Decisão (Critério: Gini)", fontsize=14)
pplt.show()
```
